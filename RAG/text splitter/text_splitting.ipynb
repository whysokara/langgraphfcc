```python
from langchain_text_splitters import CharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader
```

```python
file_path = '/Users/kara/Desktop/langgraphfcc/RAG/poem.pdf'

loader = PyPDFLoader(file_path)
docs = loader.load()
```

```python
splitter = CharacterTextSplitter(
    chunk_size = 100,
    chunk_overlap = 20,
    separator = ''
)
```

```python
result = splitter.split_documents(docs)
```

```python
result[0].page_content
```

```python
import re
data = []
for i in result:
    # join single-letter splits only
    i = re.sub(r'\b(\w)\s+(\w{2,})\b', r'\1\2', i)
    i = re.sub(r'\b(\w{2,})\s+(\w)\b', r'\1\2', i)
    i = i.replace("\xa0", " ")
    data.append(i)
```
