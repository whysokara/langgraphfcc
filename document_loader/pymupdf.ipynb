```python
import fitz
import re
```

```python
doc = fitz.open("/Users/kara/Desktop/langgraphfcc/document_loader/poem.pdf")


texts = []

for page in doc:
    text = page.get_text()

    #basic cleaning
    text = text.replace("\xa0", " ")
    text = re.sub(r"\s+", " ", text)

    texts.append(text)


full_text = "\n".join(texts)
```
